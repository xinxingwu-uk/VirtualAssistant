```python
import os
import torch
from openvoice import se_extractor
from openvoice.api import BaseSpeakerTTS, ToneColorConverter

#ckpt_base = 'checkpoints/base_speakers/ZH'
ckpt_base = 'checkpoints/base_speakers/EN'
ckpt_converter = 'checkpoints/converter'
device="cuda:0" if torch.cuda.is_available() else "cpu"
output_dir = 'outputs'

base_speaker_tts = BaseSpeakerTTS(f'{ckpt_base}/config.json', device=device)
base_speaker_tts.load_ckpt(f'{ckpt_base}/checkpoint.pth')

tone_color_converter = ToneColorConverter(f'{ckpt_converter}/config.json', device=device)
tone_color_converter.load_ckpt(f'{ckpt_converter}/checkpoint.pth')

os.makedirs(output_dir, exist_ok=True)
```

```python
#source_se = torch.load(f'{ckpt_base}/zh_default_se.pth').to(device)
source_se = torch.load(f'{ckpt_base}/en_default_se.pth').to(device)

reference_speaker = 'resources/demo_speaker1.mp3' # This is the voice you want to clone
# For recording voice https://online-voice-recorder.com/
target_se, audio_name = se_extractor.get_se(reference_speaker, tone_color_converter, target_dir='processed', vad=True)
```

```python
text="Hi everyone, I am Tucker! Welcome to our course. In this course, for video, I will try to combine manual video and AI video. It is also the first time, I have tried AI in an online course. This kind of AI is currently becoming popular, we can use AI to create video Avatar. "
```

```python
save_path = f'{output_dir}/clonedVoice_en.wav'

src_path = f'{output_dir}/tmp.wav'
base_speaker_tts.tts(text, src_path, speaker='default', language='English', speed=1.0)

# Run the tone color converter
encode_message = "@MyShell"
tone_color_converter.convert(
    audio_src_path=src_path,
    src_se=source_se,
    tgt_se=target_se,
    output_path=save_path,
    message=encode_message)
```
